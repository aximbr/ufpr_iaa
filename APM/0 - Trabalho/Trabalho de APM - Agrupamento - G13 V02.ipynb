### 1. Importar bibliotecas

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler

from sklearn.cluster import KMeans

import warnings
warnings.filterwarnings("ignore")
```

### 2. Macro definições

```python
# Bases de dados
DF_VEICULOS = "Veiculos - Dados.csv"

# Novos dados
DADOS_NOVOS_VEICULOS = "Veiculos - Novos Casos.csv"

# Seed Global (mesma utilizada nos demais experimentos do grupo -
RAND_SEED = 202613
np.random.seed(RAND_SEED)

# Numero de clusters
NUM_CLUSTERS = 10

# Largura máxima das colunas a serem exibidas
pd.set_option('display.max_colwidth', None)
```

### 3. Funções locais

```python
# Carrega DataFrame
def load_df(df_name, column_to_drop=""):
    """Função para carregar o csv e se informado, remover colunas"""
    try:
        df = pd.read_csv(df_name, sep=',')
        if column_to_drop:
            df = df.drop(column_to_drop, axis=1)
        return df
    
    except:
        print("Erro ao processar o arquivo", df_name)
```

### 4. Carrega base de dados de veiculos e faz pré-processamento

```python
scaler = StandardScaler()

df = load_df(DF_VEICULOS, ['a', 'tipo'] )

columns = list(df.columns)

X = scaler.fit_transform(df)
```

### 5. Calcula os clusters

```python

kmeans = KMeans(n_clusters=NUM_CLUSTERS, random_state=RAND_SEED)
kmeans.fit(X)

labels = kmeans.labels_
centroids = kmeans.cluster_centers_

df['cluster'] = labels

display(df.head(10))
```

```python
print('Quantidade de elementos por cluster:')
print(df['cluster'].value_counts().to_string())
```
